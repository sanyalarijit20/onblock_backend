```python
import numpy as np

def verify_identity(current_embedding, stored_embedding, threshold=0.6):
    """
    Compares two vectors using Cosine Similarity (Dot Product).
    Returns a dictionary with result and score.
    """
    # Convert lists back to numpy for math
    vec1 = np.array(current_embedding)
    vec2 = np.array(stored_embedding)
    
    # Dot product of normalized vectors
    similarity = np.dot(vec1, vec2)
    
    return {
        "verified": bool(similarity >= threshold),
        "similarity": float(similarity)
    }
```
